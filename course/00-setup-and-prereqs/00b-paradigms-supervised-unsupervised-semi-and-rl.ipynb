# ML paradigms: supervised, unsupervised, semi-supervised, reinforcement

This short notebook maps **problem types** to **what you observe** and **what you optimize**.

| Paradigm | Labels | Typical goal | Examples |
| --- | --- | --- | --- |
| **Supervised** | Every training example has a target | Predict \(y\) from \(x\) | Regression, classification |
| **Unsupervised** | No targets (or ignored) | Structure, density, representation | Clustering, PCA, anomaly detection |
| **Semi-supervised** | Labels on a subset only | Use unlabeled data to help | Label propagation, self-training |
| **Reinforcement learning** | Reward signal over time | Maximize cumulative reward | Games, control, recommender bandits |

**Not mutually exclusive in practice:** self-supervised pretraining (e.g. masked language modeling) produces representations that later feed supervised heads; bandits blur RL and supervised learning.

```python
# Tiny illustration: supervised = fit f(x) ~ y; unsupervised = no y in the loss
import numpy as np
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.cluster import KMeans

X, y = make_classification(n_samples=300, n_features=2, n_redundant=0, random_state=0)
clf = LogisticRegression().fit(X, y)
print("Supervised accuracy:", round(clf.score(X, y), 3))

km = KMeans(n_clusters=2, n_init=10, random_state=0).fit(X)
# Cluster labels are arbitrary; compare via best permutation is overkill here
print("KMeans cluster sizes:", np.bincount(km.labels_))
```
