```python
import sys
from pathlib import Path

print("Python:", sys.version)

_repo = Path.cwd().resolve()
for _ in range(12):
    if (_repo / "utils" / "data_paths.py").exists():
        sys.path.insert(0, str(_repo))
        break
    if _repo.parent == _repo:
        raise FileNotFoundError("Start Jupyter from the ml-notebook repo root.")
    _repo = _repo.parent

from utils.data_paths import data_dir, repo_root

print("Repo root:", repo_root())
print("Data dir:", data_dir())
assert (data_dir()).is_dir(), "data/ not found"
```
