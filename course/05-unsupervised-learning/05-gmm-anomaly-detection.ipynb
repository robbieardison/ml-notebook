# Gaussian mixture models and anomaly detection

**Unsupervised learning.**

- **GaussianMixture** fits a soft clustering model (each point has a probability over components).
- **IsolationForest** scores how "isolated" points are — often used for **anomaly / outlier** detection.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture
from sklearn.ensemble import IsolationForest

X, y_true = make_blobs(n_samples=400, centers=3, cluster_std=0.85, random_state=42)
gmm = GaussianMixture(n_components=3, random_state=42).fit(X)
labels = gmm.predict(X)
plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap="tab10", s=12)
plt.title("GMM cluster assignments")
plt.show()

# Inject a few outliers
rng = np.random.default_rng(0)
outliers = rng.uniform(low=-15, high=15, size=(12, 2))
X2 = np.vstack([X, outliers])
iso = IsolationForest(n_estimators=200, contamination=0.04, random_state=42)
pred = iso.fit_predict(X2)
plt.figure(figsize=(6, 5))
plt.scatter(X2[:, 0], X2[:, 1], c=np.where(pred == -1, "red", "lightgray"), s=14)
plt.title("IsolationForest: red = flagged anomaly")
plt.show()
```
