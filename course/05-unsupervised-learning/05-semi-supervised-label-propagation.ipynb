# Semi-supervised learning: label propagation

**Semi-supervised learning:** only a **fraction** of training points have labels. `LabelPropagation` diffuses labels along a similarity graph built from unlabeled data.

We mask most labels on **digits** and measure accuracy on a held-out test set.

```python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.semi_supervised import LabelPropagation
from sklearn.metrics import accuracy_score

X, y = load_digits(return_X_y=True)
X = StandardScaler().fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

rng = np.random.default_rng(42)
labeled_mask = rng.random(len(y_train)) < 0.08
y_partial = np.full_like(y_train, -1)
y_partial[labeled_mask] = y_train[labeled_mask]
print("Labeled fraction:", round(labeled_mask.mean(), 3))

lp = LabelPropagation(kernel="rbf", gamma=0.1, max_iter=500)
lp.fit(X_train, y_partial)
y_hat = lp.predict(X_test)
print("Test accuracy (label propagation):", round(accuracy_score(y_test, y_hat), 3))
```
