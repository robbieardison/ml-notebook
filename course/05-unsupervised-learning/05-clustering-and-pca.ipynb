# Clustering and PCA

- **KMeans** and **DBSCAN** on a 2D projection of digits
- **PCA** for visualization and dimensionality reduction

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA

X, y = load_digits(return_X_y=True)
X = StandardScaler().fit_transform(X)

pca = PCA(n_components=2, random_state=42)
Z = pca.fit_transform(X)

fig, ax = plt.subplots(1, 1, figsize=(7, 5))
sc = ax.scatter(Z[:, 0], Z[:, 1], c=y, cmap="tab10", s=8, alpha=0.8)
ax.set_title("PCA of digits (true labels for color)")
plt.colorbar(sc, ax=ax)
plt.show()

km = KMeans(n_clusters=10, n_init=10, random_state=42).fit(Z)
print("KMeans inertia:", round(km.inertia_, 2))

db = DBSCAN(eps=0.5, min_samples=5).fit(Z)
print("DBSCAN #labels (incl. noise -1):", len(set(db.labels_)))
print("Noise points:", np.sum(db.labels_ == -1))
```
