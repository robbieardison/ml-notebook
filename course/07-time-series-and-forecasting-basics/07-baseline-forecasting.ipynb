# Baseline time-series forecasting (classical)

We use **statsmodels** ` Holt-Winters` exponential smoothing on a synthetic series.
For real projects, validate with time-based splits and compare to naive baselines.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.holtwinters import ExponentialSmoothing

rng = np.random.default_rng(42)
t = np.arange(120)
y = 0.05 * t + 10 * np.sin(2 * np.pi * t / 12) + rng.normal(0, 0.5, size=len(t))
series = pd.Series(y, index=pd.date_range("2020-01", periods=len(t), freq="M"))

train, test = series.iloc[:-24], series.iloc[-24:]
model = ExponentialSmoothing(
    train, trend="add", seasonal="add", seasonal_periods=12
).fit(optimized=True)
fcst = model.forecast(len(test))

mae = float(np.mean(np.abs(test.values - fcst.values)))
print("MAE on holdout:", round(mae, 4))

plt.figure(figsize=(9, 4))
plt.plot(train.index, train, label="train")
plt.plot(test.index, test, label="test")
plt.plot(fcst.index, fcst, label="forecast")
plt.legend()
plt.title("Holt-Winters (example)")
plt.show()
```
