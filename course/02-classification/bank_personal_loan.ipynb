```python
import sys
from pathlib import Path

_repo = Path.cwd().resolve()
for _ in range(12):
    if (_repo / "utils" / "data_paths.py").exists():
        sys.path.insert(0, str(_repo))
        break
    if _repo.parent == _repo:
        raise FileNotFoundError("Could not find utils/data_paths.py; run Jupyter from ml-notebook repo.")
    _repo = _repo.parent

from utils.data_paths import data_dir
DATA = data_dir()

from sklearn.model_selection import train_test_split
import pandas as pd

data = pd.read_csv(DATA / "Bank_Personal_Loan_Modelling.csv")
X = data.drop(['ID', 'Personal Loan'], axis=1)
y = data['Personal Loan']

# Splitting the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## Classification Models:

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

# Initializing models
logreg_model = LogisticRegression()
knn_model = KNeighborsClassifier()
nb_model = GaussianNB()

# Training models
logreg_model.fit(X_train, y_train)
knn_model.fit(X_train, y_train)
nb_model.fit(X_train, y_train)
```

## Confusion Matrix

```python
from sklearn.metrics import confusion_matrix

# Predictions
logreg_preds = logreg_model.predict(X_test)
knn_preds = knn_model.predict(X_test)
nb_preds = nb_model.predict(X_test)

# Confusion matrices
logreg_cm = confusion_matrix(y_test, logreg_preds)
knn_cm = confusion_matrix(y_test, knn_preds)
nb_cm = confusion_matrix(y_test, nb_preds)

print("Logistic Regression Confusion Matrix:")
print(logreg_cm)
print("\nK-NN Confusion Matrix:")
print(knn_cm)
print("\nNaïve Bayes Confusion Matrix:")
print(nb_cm)
```
