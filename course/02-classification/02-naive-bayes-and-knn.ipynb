# Naive Bayes and k-Nearest Neighbors

**Supervised learning — classification.** Two classical, complementary baselines:

- **Gaussian Naive Bayes:** fast, probabilistic, strong when features are roughly Gaussian per class.
- **k-NN:** non-parametric, uses local neighborhoods; sensitive to scale (we standardize).

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report

data = load_wine()
X, y = data.data, data.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

gnb = Pipeline([("scale", StandardScaler()), ("clf", GaussianNB())])
knn = Pipeline([("scale", StandardScaler()), ("clf", KNeighborsClassifier(n_neighbors=7))])

for name, model in [("GaussianNB", gnb), ("kNN_k=7", knn)]:
    model.fit(X_train, y_train)
    print("===", name, "===")
    print(classification_report(y_test, model.predict(X_test), target_names=data.target_names))
    cv = cross_val_score(model, X_train, y_train, cv=5)
    print("CV accuracy mean:", round(cv.mean(), 3), "std:", round(cv.std(), 3))
```
